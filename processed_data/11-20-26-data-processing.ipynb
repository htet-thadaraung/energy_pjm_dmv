```r
#loaded libraries needed for reading excel folders and data cleaning. 

library(readxl)
library(tidyr)
library(dplyr)
```

```r
# Before loading in xlsx files from the raw data, I used excel_sheets to highlight all the individual sheets listed within the excel file. 
excel_sheets("/Users/rufous-piculet/Desktop/energy_pjm_dmv/raw_data/net_metering2024_formEIA-861M.xlsx")

#Sheets will be loaded in individually. File path created, and necessary information for analysis is loaded in. Territory information sheet (Puerto Rico) not loaded. 
file_path_2024 <- "/Users/rufous-piculet/Desktop/energy_pjm_dmv/raw_data/net_metering2024_formEIA-861M.xlsx"

# Skipped row containing sheet name. Col names is set to FALSE so that the merged items are not saved as headers. 
utility_level_states_2024 <- read_excel(file_path_2024, sheet = "Utility_Level-States", skip = 1, col_names = FALSE)
utility_state_TPO_2024 <- read_excel (file_path_2024, sheet = "Utility_Total-TPO", skip = 1, col_names =  FALSE)
monthly_totals_states_2024 <- read_excel (file_path_2024, sheet = "Monthly_Totals-States", skip = 1, col_names = FALSE)
```

```r
# The first three rows have to be combined and translated into corresponding columns and their data for cleaning. Created temporary rows so that the information can be cleaned up. 
row1 <- tibble(x = as.character(utility_level_states_2024[1, ])) %>%
  fill(x, .direction = "down") %>%
  pull(x)

row2 <- tibble(x = as.character(utility_level_states_2024[2, ])) %>%
  fill(x, .direction = "down") %>%
  pull(x)

row3 <- as.character(utility_level_states_2024[3, ])

#All three rows are then combined. NAs were ignored. ID columns stayed the same.
new_names <- sapply(seq_along(row3), function(i) {
  parts <- c(row1[i], row2[i], row3[i])
  parts <- parts[!is.na(parts) & parts != "Utility Characteristics"]
  paste(parts, collapse = "_")
})

#All three rows were dropped, and new column names were added accordingly. 
utility_level_states_2024_clean <- utility_level_states_2024[-c(1:3), ]
names(utility_level_states_2024_clean) <- new_names

# Checked the first 10 column names to see that the new column names and the data makes sense. 
names(utility_level_states_2024_clean)[1:10]
head(utility_level_states_2024_clean)
```

```r
#Repeating the process for state TPO sheet. 
row1 <- tibble(x = as.character(utility_state_TPO_2024[1, ])) %>%
  fill(x, .direction = "down") %>%
  pull(x)

row2 <- tibble(x = as.character(utility_state_TPO_2024[2, ])) %>%
  fill(x, .direction = "down") %>%
  pull(x)

row3 <- as.character(utility_state_TPO_2024[3, ])

new_names <- sapply(seq_along(row3), function(i) {
  parts <- c(row1[i], row2[i], row3[i])
  parts <- parts[!is.na(parts) & parts != "Utility Characteristics"]
  paste(parts, collapse = "_")
})

utility_state_TPO_2024_clean <- utility_state_TPO_2024[-c(1:3), ]
names(utility_state_TPO_2024_clean) <- new_names

names(utility_state_TPO_2024_clean)[1:10]
head(utility_state_TPO_2024_clean)
```
